The headline number is **`range / d_i`**: how far a swept-up upstream Al ion travels before
Coulomb collisions stop it, in units of the ion inertial length the shock transitions over.
Much greater than 1 means collisions cannot have built the ramp.

## Scripted use

The panel is only a front-end; the physics is importable on its own for scans and tables.

```python
import numpy as np
from magshockz.common.collisionality import shock_collisionality

for v_km_s in [200, 292, 500, 919]:
    r = shock_collisionality(v_inflow=v_km_s * 1e3, n_i=1e18 * 1e6,
                             T_i=30.0, T_e=30.0, z=5.0, b=20.0)
    print(f"v={v_km_s:4d} km/s   range={r.ion.stopping_range*1e6:9.3g} um   "
          f"d_i={r.d_i*1e6:6.1f} um   range/d_i={r.knudsen:7.2f}")
```
